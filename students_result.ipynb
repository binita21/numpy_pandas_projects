## Creating student info 

```python
import pandas as pd
import numpy as np
student_info = pd.DataFrame({
    'studentid':[101,102,103,104,105,106,107,108],
'Name':['Aayush','Ali','Binisha','Bijeta','Chandani','Chandra','Heena','Raju'],
'Age':[19,18,19,19,18,17,18,18],
'Math':[85,92,70,77,80,23,90,50],
'Science':[78,20,75,80,77,56,65,80],
'English':[85,80,81,90,91,90,89,76],
'Computer':[88,89,55,80,34,56,76,60],
'Social':[72,76,80,91,77,80,98,44],
'Attendance':[95,98,90,90,55,95,90,60]
})
```

##### converted to csv file

```python
student_info.to_csv('student_info.csv')
```

```python
student_info
```

#### Adding new column Gender

```python
student_info['Gender'] = ['Male','Male','Female','Female','Female','Male','Female','Male']
```

```python
student_info
```

#### Computing and adding the Total to the student_info dataframe

```python
# Computing and adding the Total to the student_info dataframe
student_info['Total']= student_info['Math']+student_info['Science']+student_info['English']+student_info['Computer']+student_info['Social']
```

```python
student_info
```

#### Computing Total Percentage obtained by each student

```python
no_of_subjects=5
student_info['Percentage']=student_info['Total']/no_of_subjects
```

```python
student_info
```

## Calculate the Gpa according to Percentage and display the Gpa

```python
print(student_info.columns)
```

```python
student_info['Grade'] =student_info.loc[(student_info['Percentage'] >= 90), 'Grade'] = 'A+'
student_info.loc[(student_info['Percentage'] >= 80) & (student_info['Percentage'] < 90), 'Grade'] = 'A'
student_info.loc[(student_info['Percentage'] >= 70) & (student_info['Percentage'] < 80), 'Grade'] = 'B'
student_info.loc[(student_info['Percentage'] >= 60) & (student_info['Percentage'] < 70), 'Grade'] = 'C'
student_info.loc[(student_info['Percentage'] < 60), 'Grade'] = 'F'
```

```python
student_info
```

### Creating a condition where student who got below 40 in any subject fails and who got above or equal to 40 will pass

```python
student_info['Result']=student_info.loc[(student_info['Math'] >= 40) & (student_info['Science'] >= 40) & (student_info['English'] >= 40) & (student_info['Computer'] >= 40) & (student_info['Social'] >= 40) , 'Result']='PASS'
student_info.loc[(student_info['Math'] < 40) | (student_info['Science'] < 40) | (student_info['English'] < 40) | (student_info['Computer'] < 40) | (student_info['Social'] < 40) , 'Result']='FAIL'
```

```python
student_info
```

```python
student_info.groupby('Result')['Result'].count()
```

## Finding the maximum percentage obtained with all details

```python
# Here we can see the rank according to the percentage (i.e. topper who score highest to lowest)
student_info.groupby('Percentage').max().sort_values('Percentage',ascending=False)
```

```python
# Finding the highest marks obtained in each subject
print('Highest marks scored in Maths is ',student_info["Math"].max())
print('Highest marks scored in Science is ',student_info["Science"].max())
print('Highest marks scored in English is ',student_info["English"].max())
print('Highest marks scored in Computer is ',student_info["Computer"].max())
print('Highest marks scored in Social is ',student_info["Social"].max())
```

## Finding the average marks obtained in each subject

```python

print('Average marks scored in Maths is ',student_info["Math"].mean())
print('Average marks scored in Science is ',student_info["Science"].mean())
print('Average marks scored in English is ',student_info["English"].mean())
print('Average marks scored in Computer is ',student_info["Computer"].mean())
print('Average marks scored in Social is ',student_info["Social"].mean())
```
